## 2 · Get the made-up sample data (1.7 MB, from GitHub)

```python
!git clone -q --depth 1 --filter=blob:none --sparse https://github.com/wealthtechnical-afk/nse-bse-eod-downloader.git wt_kit > /dev/null 2>&1
!cd wt_kit && git sparse-checkout set india-data-kit/sample_data > /dev/null 2>&1
DATA = "wt_kit/india-data-kit/sample_data/NSE_EOD"   # on your PC: wti.Kit() finds the Downloader's folder by itself
print("sample data ready:", DATA)
```

## 3 · Load a whole watchlist in one line

```python
import pandas as pd, wt_india_eod as wti
kit = wti.Kit(DATA)
print("stocks:", len(kit.symbols()), "| watchlists:", ", ".join(kit.watchlists()))
df = kit.stocks("DEMO40")          # with real data: kit.stocks("NIFTY500")
print(df.shape, "rows ·", df.symbol.nunique(), "stocks ·", df.date.min().date(), "→", df.date.max().date())
df.tail()
```

## 4 · Market breadth

```python
b = wti.breadth(df)
b.tail()
```

```python
ax = b[["pct_above_50dma", "pct_above_200dma"]].plot(figsize=(10, 4), title="% of stocks above their 50/200-day average (sample data)")
ax.set_ylim(0, 100); ax.axhline(50, ls="--", lw=1, color="grey");
```

## 5 · Sector rotation

```python
kit.sector_rotation()
```

## 6 · Data-quality checks

```python
issues = wti.check_quality(df)
print(len(issues), "possible issues")
issues.head(10)
```

## 7 · Export to other platforms

```python
import os
for fmt in ["amibroker", "nt8", "metastock-csv", "mt5"]:
    wti.WRITERS[fmt](df, f"out/{fmt}")
    print(f"{fmt:14s}", len(os.listdir(f"out/{fmt}")), "files")
wti.to_sqlite(df, "out/eod.sqlite"); print("sqlite        ", os.path.getsize("out/eod.sqlite"), "bytes")
```
